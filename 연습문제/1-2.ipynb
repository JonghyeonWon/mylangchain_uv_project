```python
import os
from dotenv import load_dotenv

load_dotenv()
OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

if not OPENAI_API_KEY:
    raise RuntimeError(".env 파일에 OPENAI_API_KEY 가 없습니다.")
print("OPENAI_API_KEY 설정 확인 완료")
```

```python
from langchain_openai import ChatOpenAI

OPENAI_MODEL = "gpt-4o-mini" 
TEMPERATURE = 0.7

def get_llm(model: str = OPENAI_MODEL, temperature: float = TEMPERATURE) -> ChatOpenAI:
    return ChatOpenAI(
        api_key=OPENAI_API_KEY,
        model=model,
        temperature=temperature,
    )

llm = get_llm()
```

```python
def print_place(place):
    print(f"1단계 결과: {place}")
    print("\n")
    return place
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

prompt1 = ChatPromptTemplate.from_messages([
    ("system", "당신은 여행 전문가입니다. 사용자가 특정 도시나 국가를 입력하면 해당 지역의 대표적인 관광 명소를 추천해주되 명소의 이름만 답변하세요."),
    ("user", "{destination}의 대표적인 관광 명소 1가지를 추천해주세요.")
])

prompt2 = ChatPromptTemplate.from_messages([
    ("system", "당신은 관광 명소에 대한 상세 정보를 알려주는 가이드입니다."),
    ("user", """{place}에 대한 상세 정보를 아래 형식으로 알려주세요.

    명소: {place}
    역사:
    특징:
    방문 팁:
""")
])

chain1 = prompt1 | llm | StrOutputParser()

chain2 = (
    {"place": chain1 | print_place}
    | prompt2
    | llm
    | StrOutputParser()
)

destination = "로마"
print(f"입력: {destination}")

response = chain2.invoke({"destination": destination}) 

print(f"2단계 결과: {response}")
```
