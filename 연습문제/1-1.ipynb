```python
import os
from dotenv import load_dotenv

load_dotenv()
OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

if not OPENAI_API_KEY:
    raise RuntimeError(".env 파일에 OPENAI_API_KEY 가 없습니다.")
print("OPENAI_API_KEY 설정 확인 완료")
```

```python
from langchain_openai import ChatOpenAI

OPENAI_MODEL = "gpt-4o-mini" 
TEMPERATURE = 0.7

def get_llm(model: str = OPENAI_MODEL, temperature: float = TEMPERATURE) -> ChatOpenAI:
    return ChatOpenAI(
        api_key=OPENAI_API_KEY,
        model=model,
        temperature=temperature,
    )

llm = get_llm()
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

prompt = PromptTemplate.from_template(
    """당신은 AI 요리사입니다. 
    사용자가 재료를 입력하면 그 재료로 만들 수 있는 요리 1가지를 추천해주고
    간단한 레시피를 알려주세요.

    <재료>: {ingredients}
    
    아래 형식으로 답변하세요.
    추천 요리:
    재료:
    조리법:
    1.
    2.
    3.
    ...
"""
)

chain = prompt | llm | StrOutputParser()

ingredients = "계란, 밥, 김치"
print(f"입력: {ingredients}")
answer = chain.stream({"ingredients": ingredients})

for token in answer:
    print(token, end="", flush=True)
```
